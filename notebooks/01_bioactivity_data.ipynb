## Fetch, clean, and label IC50 data

Steps inside `run_part1()`:

1. Download IC50 records for CHEMBL220
2. Drop missing `standard_value` / `canonical_smiles`
3. Keep nM units only
4. Drop duplicate SMILES
5. Label **active** (≤ 1000 nM), **inactive** (≥ 10,000 nM), **intermediate**

The first run talks to the ChEMBL API and can take several minutes. Re-run with `skip_fetch=True` to reuse `data/raw/`.

````python
curated = run_part1(skip_fetch=False)
curated.head()
````
